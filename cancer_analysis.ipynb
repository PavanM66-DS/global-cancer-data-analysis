```python
fig, axes = plt.subplots(2,2, figsize=(15,10))

# 1. Gender distribution
p1 = sn.countplot(data=d, x='Sex', ax=axes[0,0])
axes[0, 0].set_title('Patient Records by Gender')
axes[0, 0].set_xlabel('Gender')
axes[0, 0].set_ylabel('Number of Records')
for i in p1.containers:
  p1.bar_label(i)

# 2. Cancer type distribution

p2 = sn.countplot(
    data=d,
    x='Cancer_Type',
    order=d['Cancer_Type'].value_counts().index,
    ax=axes[0, 1]
)
axes[0, 1].set_title('Patient Records by Cancer Type')
axes[0, 1].set_xlabel('Number of Records')
axes[0, 1].set_ylabel('Cancer Type')
for j in p2.containers:
  p2.bar_label(j)

# 3. Age distribution
p3 = sn.histplot(data=d, x='Age', bins=10, ax=axes[1,0], multiple='dodge')
axes[1, 0].set_title('Age Distribution')
axes[1, 0].set_xlabel('Age')
axes[1, 0].set_ylabel('Frequency')

for k in p3.containers:
  p3.bar_label(k)

# 4. Treatment cost distribution
sn.histplot(
    data=d,
    x='Treatment_Cost_USD',
    bins=10,
    ax=axes[1, 1]
)
axes[1, 1].set_title('Treatment Cost Distribution')
axes[1, 1].set_xlabel('Treatment Cost (USD)')
axes[1, 1].set_ylabel('Frequency')

plt.suptitle('Global Cancer Data — Exploratory Analysis', fontsize=16)
plt.tight_layout()
plt.show()
```
